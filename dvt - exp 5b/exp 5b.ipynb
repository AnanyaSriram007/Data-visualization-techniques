```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(r"C:\Users\Anusri\Downloads\breast-cancer.csv")

if "Unnamed: 32" in df.columns:
    df = df.drop("Unnamed: 32", axis=1)

print(df.head())
```

```python
plt.figure(figsize=(7,5))

plt.plot(df.index[:30], df["radius_mean"][:30])

plt.title("Mean Radius of First 30 Patients")
plt.xlabel("Patient Index")
plt.ylabel("Mean Radius")

plt.show()
```

```python
avg_radius = df.groupby("diagnosis")["radius_mean"].mean()

plt.figure(figsize=(7,5))

plt.bar(avg_radius.index, avg_radius.values)

plt.title("Average Radius by Diagnosis")
plt.xlabel("Diagnosis")
plt.ylabel("Average Radius")

plt.show()
```

```python
diagnosis_count = df["diagnosis"].value_counts()

plt.figure(figsize=(6,6))

plt.pie(
    diagnosis_count.values,
    labels=diagnosis_count.index,
    autopct="%1.1f%%"
)

plt.title("Breast Cancer Diagnosis Distribution")

plt.show()
```

```python
plt.figure(figsize=(7,5))

plt.hist(df["radius_mean"], bins=20)

plt.title("Distribution of Mean Radius")
plt.xlabel("Mean Radius")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(7,5))

plt.scatter(
    df["radius_mean"],
    df["area_mean"]
)

plt.title("Mean Radius vs Mean Area")
plt.xlabel("Mean Radius")
plt.ylabel("Mean Area")

plt.show()
```

```python
plt.figure(figsize=(6,5))

plt.boxplot(df["area_mean"])

plt.title("Box Plot of Mean Area")
plt.ylabel("Mean Area")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.plot(
    df.index[:20],
    df["radius_mean"][:20],
    label="Mean Radius"
)

plt.plot(
    df.index[:20],
    df["texture_mean"][:20],
    label="Mean Texture"
)

plt.title("Radius and Texture Comparison")
plt.xlabel("Patient Index")
plt.ylabel("Measurement")

plt.legend()

plt.show()
```
