```python
import pandas as pd
import numpy as np

data = {
    "EmpID": [101, 102, 103, 104, 104, 105, 106, 107, 108, 109,
              110, 111, 112, 113, 114, 115, 116, 117, 118, 119],

    "Name": ["Arun", "Bala", "Divya", "Priya", "Priya", "Ravi",
             "Kiran", "Meena", "Suresh", "Anitha",
             "Vijay", "Sneha", "Rahul", "Deepa", "Kumar",
             "Lakshmi", "Ajay", "Nisha", "Gokul", "Pooja"],

    "Age": [25, np.nan, 28, 25, 25, 60,
            35, 42, np.nan, 29,
            31, 26, 40, 38, 45,
            np.nan, 30, 27, 50, 33],

    "Salary": [25000, 30000, np.nan, 25000, 25000, 150000,
               45000, 52000, 40000, np.nan,
               55000, 32000, 70000, 68000, 80000,
               48000, 50000, 36000, np.nan, 62000],

    "Department": ["HR", "IT", "HR", "HR", "HR", "Finance",
                   "IT", "Marketing", "Finance", "HR",
                   "IT", "Marketing", "Finance", "HR", "IT",
                   "Finance", "Marketing", "HR", "IT", "Finance"]
}

df = pd.DataFrame(data)

print(df)

df.to_csv("employee_data.csv", index=False)

print("\nDataset saved as employee_data.csv")
```

```python
print(df.isnull())
print(df.isnull().sum())
```

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())

df["Salary"] = df["Salary"].fillna(df["Salary"].mean())

print(df)
```

```python
print(df.duplicated())
print("Duplicate rows=",df.duplicated().sum())
df.drop_duplicates(inplace=True)
print(df)
```

```python
Q1 = df["Salary"].quantile(0.25)
Q3 = df["Salary"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df["Salary"] < lower) | (df["Salary"] > upper)]

print(outliers)
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df[["Salary"]] = scaler.fit_transform(df[["Salary"]])

print(df[["Salary"]])
```

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

df["Department"] = encoder.fit_transform(df["Department"])

print(df)
```

```python
one_hot = pd.get_dummies(df["Department"])

print(one_hot)
```

```python
df.to_csv("cleaned_employee_data.csv", index=False)

print("Dataset saved successfully.")
```
