```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Patient": ["P1", "P2", "P3", "P4", "P5",
                "P6", "P7", "P8", "P9", "P10"],

    "Diagnosis": ["B", "M", "B", "M", "B",
                  "M", "B", "M", "B", "M"],

    "Radius": [12.5, 18.2, 11.8, 20.1, 13.2,
               17.5, 10.9, 21.3, 14.1, 19.0],

    "Texture": [18.4, 22.1, 16.2, 25.3, 19.5,
                23.4, 15.8, 26.1, 20.2, 24.5],

    "Area": [490, 1040, 430, 1250, 540,
             980, 390, 1400, 610, 1150]
}

df = pd.DataFrame(data)

print(df)
```

```python
plt.plot(df["Patient"], df["Radius"])

plt.title("Tumor Radius by Patient")
plt.xlabel("Patient")
plt.ylabel("Radius")

plt.show()
```

```python
avg_radius = df.groupby("Diagnosis")["Radius"].mean()

plt.bar(avg_radius.index, avg_radius.values)

plt.title("Average Tumor Radius by Diagnosis")
plt.xlabel("Diagnosis")
plt.ylabel("Average Radius")

plt.show()
```

```python
diagnosis_count = df["Diagnosis"].value_counts()

plt.pie(
    diagnosis_count.values,
    labels=diagnosis_count.index,
    autopct="%1.1f%%"
)

plt.title("Breast Cancer Diagnosis Distribution")

plt.show()
```

```python
plt.hist(df["Radius"], bins=5)

plt.title("Distribution of Tumor Radius")
plt.xlabel("Radius")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.scatter(df["Radius"], df["Area"])

plt.title("Tumor Radius vs Area")
plt.xlabel("Radius")
plt.ylabel("Area")

plt.show()
```

```python
plt.boxplot(df["Area"])

plt.title("Box Plot of Tumor Area")
plt.ylabel("Area")

plt.show()
```
